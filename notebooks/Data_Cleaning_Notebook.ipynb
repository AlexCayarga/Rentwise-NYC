```python
mixed_type_found = False

for col in df.columns:
    types = df[col].map(type).value_counts()

    if len(types) > 1:
        mixed_type_found = True
        print("Mixed types in:", col)
        print(types)
        print()

if not mixed_type_found:
    print("No mixed-type columns found.")
```
